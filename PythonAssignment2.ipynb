## Data Structures - List, Dictionary, Set and Conditional Statements

1. List (Creation, Modification and Access)
1(a) Create age_list with five integer elements

```python
age_list = [24, 25, 26, 27, 28]
print(age_list)
```

1(b) Create name_list with five string elements

```python
name_list = ["Anu", "Akhil", "Meera", "Arun", "Diya"]
print(name_list)
```

List Operations / Modifications

```python
# a. Append "Yazhini" to name_list
name_list.append("Yazhini")
print("After append:", name_list)

# b. Insert 30 at index 2 in age_list
age_list.insert(2, 30)
print("After insert:", age_list)

# c. Remove "Yazhini" from name_list
name_list.remove("Yazhini")
print("After remove:", name_list)

# d. Pop the last element from age_list
age_list.pop()
print("After pop:", age_list)

# e. Extend age_list with additional ages
age_list.extend([29, 30, 26])
print("After extend:", age_list)

# f. Sort age_list in descending order
age_list.sort(reverse=True)
print("Descending order:", age_list)

# g. Find maximum age, minimum age and sum
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

3. Accessing List Elements

```python
print("First element:", name_list[0])
print("Last element:", name_list[-1])
print("Elements from index 2 to 4:", name_list[2:5])
print("Reverse order:", name_list[::-1])
```

Dictionary Creation

```python
student_marks = {
    "Anu": 85,
    "Akhil": 78,
    "Meera": 92,
    "Arun": 75,
    "Diya": 88
}

print(student_marks)
```

Access a Student's Mark

```python
print("Meera's mark:", student_marks["Meera"])
```

Add a New Student

```python
student_marks["Janani"] = 80
print(student_marks)
```

Update a Student's Mark

```python
student_marks["Akhil"] = 82
print(student_marks)
```

Display Keys, Values and Items

```python
print("Student names:", student_marks.keys())
print("Student marks:", student_marks.values())
print("Student details:", student_marks.items())
```

Set Creation and Duplicate Analysis

```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print(my_set)
```

Changing a Set Element

```python
my_set = {'a', 'e', 'i', 'o', 'u'}

try:
    my_set[4] = 's'
except TypeError as error:
    print("Error:", error)
```

To add an element to the set, use:

```python
my_set.add('s')
print(my_set)
```

Create Two Sets

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)
```

Union and Intersection

```python
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

Performance Category Program
